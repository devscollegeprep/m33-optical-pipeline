```python
from astropy.io import fits
```

* Maps the standard `fits` binary streaming decoder directly into the local environment
* Binds externa; binary streaming structures into a localized tool

The math: Establishes the astronomical file parsing operator function $f_{\text{fits}}$ to map linear text string paths to structured binary multi-matrix data blocks
$$f_{\text{fits}}(x) = y \quad \text{where} \quad x \in \text{Paths}, \,\, y \in \text{HDULists}$$

```python
simulated_hdu = fits.PrimaryHDU()
simulated_hdu.header["IMAGETYP"] = "LIGHT"
simulated_hdu.writeto("../data/raw/simulated_frame.fits", overwrite=True)
print("FITS Simulation Engine complete. Core metadata target generated on disk.")
```

* Creates an empty primary header data unit (HDU) object, creates a custom header string. and writes the binary stream to disk storage (essentially creating a fake .fits data set to test coding lines)
* Instantiates an Astropy HDU database, maps a string key to a string value inside its header dictionary, and exports a raw `/fits` file matrix

```python
target_file_path = "../data/raw/simulated_frame.fits"
with fits.open(target_file_path) as hdul:
    file_header = hdul[0].header
    frame_type = file_header.get("IMAGETYP", "UNKNOWN")
print(f"decoded telemetry badge - frame type: {frame_type}")
```

* Depolys an automated text manager that looks deep into the file names and extracts the exact specific type of .fits file string attatched
* Maps a binary stream database object to a local memory header block and returns a clean string to the window

The math: Formally defines the header evaluation operator $H$ which acquires a binary FITS structure $x$ and a key string $k$, mapping them to a finite scalar token space $mathcal{T}$:
$$H(x, k) = v \quad \text{where} \quad x \in \mathcal{F}, \,\, k = \text{"IMAGETYP"}$$

Where $\mathcal{F}$ represents the database of raw telescope files on disk, and the outputted string value identifies boundedly within our classification set:

$$v \in \{\text{"LIGHT"}, \text{"DARK"}, \text{"FLAT"}, \text{"BIAS"}\}$$

```python
with fits.open(target_file_path) as hdul:
    file_header = hdul[0].header
    exposure_seconds = float(file_header.get("EXPTIME", 0.0))
print(f"shutter telemetry verified - exposure time: {exposure_seconds}")
```

* Re-opens the FITS data framework to target and parse the exact electronic shutter duration of each file as a float value
* Maps a dictionary header value directly to a localized numerical decimal slot (float)

The math: Executes the header lookup operator $H$ to extract the baseline exposure interval $\Delta t$ from the file vector matrix $x$:
$$H(x, \text{"EXPTIME"}) = \Delta t \quad \text{where} \quad \Delta t \in \mathbb{R}^+$$
