```python
target_folder_files = os.listdir("../data/raw")
raw_lights_vault = []
raw_darks_vault = []
raw_flats_vault = []
raw_biases_vault = []
for current_file in target_folder_files:
    full_target_path = os.path.join("../data/raw", current_file)
    with fits.open(full_target_path) as hdul:
        current_header = hdul[0].header
        frame_classification = current_header.get("IMAGETYP", "UNKNOWN").upper()
        raw_pixel_array = hdul[0].data.copy() if hdul[0].data is not None else np.zeros((10,10))
        sky_background_median = np.median(raw_pixel_array)
        if frame_classification == sky_background_median < 100.0:
            print(f" CLOUD SHIELD INTERVENE: Frame dropped: {sky_background_median}")
        if frame_classification == "LIGHT":
            raw_pixel_array = np.ascontiguousarray(raw_pixel_array, dtype=np.float32)
        if frame_classification == "LIGHT":
            pixel_contrast_spread = np.std(raw_pixel_array)
            if pixel_contrast_spread < 15.0:
                print(f"BLUR DETECTED -> frame dropped due to wind shake: {pixel_contrast_spread}px")
            continue
        if "LIGHT" in frame_classification:
            raw_lights_vault.append(raw_pixel_array)
        elif "DARK" in frame_classification:
            raw_darks_vault.append(raw_pixel_array)
        elif "FLAT" in frame_classification:
            raw_flats_vault.append(raw_pixel_array)
        elif "BIAS" in frame_classification:
            raw_biases_vault.append(raw_pixel_array)
print(f"sorting inventory complete -> Lights: {len(raw_lights_vault)} Darks: {len(raw_darks_vault)} | Flats: {len(raw_flats_vault)} | Biases: {len(raw_biases_vault)}")
```

* Effectively maps hardware path into raw data folder, created specialized vaults for each fits frame image type, acquires IMAGETYP string, creates multiple safety shields (one to protect hardward RAM overload, one to prevent cloud-sabotaged frames from being stacked and accounted for noise, one to apply base value to prevent attribute error to NoneType file, one to average out hig contrasted singular pixel cosmic rays, and one adjust blur based on wind shaking) and classifies each image based off of their IMAGETYP string to be stacked into master arrays

```python
from astropy.stats import sigma_clip
if len(raw_lights_vault) > 0:
    lights_stack = np.array(raw_lights_vault)
    clipped_stack = sigma_clip(lights_stack, sigma=3, axis=0)
    master_light = np.mean(clipped_stack, axis = 0)
else:
    master_light = np.zeros((10, 10))
if len(raw_darks_vault) > 0:
    master_dark = np.mean(raw_darks_vault, axis=0)
else:
    master_dark = np.zeros((10, 10))
if len(raw_biases_vault) > 0:
    master_bias = np.mean(raw_biases_vault, axis=0)
else:
    master_bias = np.ones((10, 10))
if len(raw_flats_vault) > 0:
    master_flat = np.mean(raw_flats_vault, axis=0)
else:
    master_flat = np.ones((10, 10))
```

* Systematically aligns classified fits images from subsected vaults using exact pixel coordinates and averages out noise values (floats) to create master stacked arrays
* Light specific, creates a safety gate for possible satelite standard deviation increases to average out into uniform sky
* Structures a safety gate for corrupted files with base pixel grid values in order to not crash production scripts

The math: Statistical compilation of noise values in individual sensor coordinate matrices into master mean value frames is compiled using the following linear equations:
$$\mathbf{M}_{\text{dark}} = \frac{1}{N} \sum_{i=1}^{N} \mathbf{D}_i \quad \text{where} \quad \mathbf{D} \in \mathbb{R}^{10 \times 10}$$

$$\mathbf{M}_{\text{bias}} = \frac{1}{K} \sum_{j=1}^{K} \mathbf{B}_j \quad \text{where} \quad \mathbf{B} \in \mathbb{R}^{10 \times 10}$$

$$\mathbf{M}_{\text{flat}} = \frac{1}{M} \sum_{k=1}^{M} \mathbf{F}_k \quad \text{where} \quad \mathbf{F} \in \mathbb{R}^{10 \times 10}$$

$$\mathbf{M}_{\text{light}} = \frac{1}{P} \sum_{m=1}^{P} \mathbf{L}_m \quad \text{where} \quad \mathbf{L} \in \mathbb{R}^{10 \times 10}$$
Taking the overall sum of all aligning pixel values and calculatiing a mean to lower calibrated image noise.

```python
denominator_field = master_flat - master_bias
safe_denominator = np.clip(denominator_field, 0.0001, None)
calibrated_master_image = (master_light - master_dark) / (safe_denominator)
print(f"PIXEL MATRIX REDUCTION COMPLETE")
print(f"pristine calibrated matrix dimensions: {calibrated_master_image.shape}")
```

* Compiles the denominator field by executing matrix subtraction between the master flat field and master bias noise layers
* Crates a safety shield using NumPy clip to insure the denominator values at a minimum baseline boundary of 0.0001, preventing fatal zero-division errors across millions of pixel noise values
* Executes the master astronomical data reduction formula concurrently across all aligned multi-dimensional coordinate grids simultaneously to output a pristine calibrated image matrix

The math: The final noise reduction and sensor noise calibration matrix transformation is executed using the following vectorized algebraic expression:

$$\mathbf{C} = \frac{\mathbf{M}_{\text{light}} - \mathbf{M}_{\text{dark}}}{\max(\mathbf{M}_{\text{flat}} - \mathbf{M}_{\text{bias}}, \, 0.0001)} \quad \text{where} \quad \mathbf{C} \in \mathbb{R}^{10 \times 10}$$

Extracting the light viable data by removing out dark thermal noise from the numerator, while utilizing the clipped denominator field to digitally flatten dust artifacts and electronic noise symmetrically across the exact row-and-column layout of the image.
